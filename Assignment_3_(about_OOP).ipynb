Q2.Write a Python program demonstrating single, multilevel, and hierarchical inheritance using simple classes such as Person, Student, and Teacher. Include at least one overridden method in the child classes to show method overriding.

```python
class person:
  def __init__(self,name):self.name= name
  def show(self):
      print("i am a person")
class student(person):
   def show(self):
      print("i am a student")
class graduatestudent(student):
   def study(self):
      print(self.name, "ia studing at graduate level.")
class teacher(person):
   def show(self):
      print("i am a teacher")
s1 = student("Rokeya")
s1.show()

g1 = graduatestudent("Nipa")
g1.show()
g1.study()

t1 = teacher("Mr. Rana")
t1.show()
```

Q3.Write a Python program to demonstrate encapsulation using private attributes with getter and setter methods. Also demonstrate method overloading and multiple inheritance using suitable classes and objects.

```python
# Encapsulation
class Account:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, balance):
        self.__balance = balance


    def display(self, message="Account Details"):
        print(message)
        print("Name:", self.name)
        print("Balance:", self.__balance)



a1 = Account("Rokeya", 5000)

a1.display()

print()


print("Balance:", a1.get_balance())


a1.set_balance(7000)

print("Updated Balance:", a1.get_balance())

print()


a1.display()
a1.display("Updated Account Details")


class Father:
    def father_method(self):
        print("This is Father's method.")


class Mother:
    def mother_method(self):
        print("This is Mother's method.")


class Child(Father, Mother):
    def child_method(self):
        print("This is Child's method.")



c1 = Child()

c1.father_method()
c1.mother_method()
c1.child_method()
```
