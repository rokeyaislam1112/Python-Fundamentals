Write a python program that takes a number as input and checks whether it is even or odd

```python
num = int(input("Enter a number: "))
if num % 2 == 0:
      print("Even")
else:
      print("odd")
```

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter oparetor(+,-,*,/):")
if op == '+':
      print(a+b)
elif op == '-':
      print(a-b)
elif op == '*':
      print(a*b)
elif op == '/':
      print(a/b)



```

3) Write a program using a for loop that calculate the sum of even number between 1 and 100

```python
sum = 0
for i in range(2 , 101 , 2):
    sum += i
print("sum of even number:", sum)
```
